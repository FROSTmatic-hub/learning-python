Practice Problem: Online Store Order Dispatcher
Scenario
You are managing an online store's dispatch system. You have a list of orders. Each order is represented as a dictionary containing the item name, customer membership status, and payment confirmation status. You need to loop through the orders and decide whether to Ship Immediately, Hold Order, or Cancel Order.

Given Starter Code
Copy this starter code directly into your script:

Python
# List of Orders
orders = [
    {"item": "Headphones", "is_paid": True, "is_vip": False},
    {"item": "Keyboard", "is_paid": False, "is_vip": True},
    {"item": "Mousepad", "is_paid": True, "is_vip": True},
    {"item": "Monitor", "is_paid": False, "is_vip": False}
]

# Output Buckets
ship_now = []
hold_payment = []
canceled = []
Tasks & Instructions
Write a for loop that iterates through each order in orders. Inside the loop, construct an if-elif-else block:

Condition 1 (if): Ready to Ship

Check if order["is_paid"] is truthy (True).

If true, append order["item"] to the ship_now list.

Condition 2 (elif): VIP Unpaid Hold

Check if order["is_paid"] is falsy (False) AND order["is_vip"] is truthy (True).

If true, append order["item"] to the hold_payment list.

Condition 3 (else): Non-VIP Unpaid Cancellation

Otherwise (unpaid non-VIPs), append order["item"] to the canceled list.

Output:

After the loop finishes, print all three destination lists (ship_now, hold_payment, canceled).

```python
# List of Orders
orders = [
    {"item": "Headphones", "is_paid": True, "is_vip": False},
    {"item": "Keyboard", "is_paid": False, "is_vip": True},
    {"item": "Mousepad", "is_paid": True, "is_vip": True},
    {"item": "Monitor", "is_paid": False, "is_vip": False}
]

# Output Buckets
ship_now = []
hold_payment = []
canceled = []
for order in orders:
    if order["is_paid"]:
        ship_now.append(order["item"])
    elif not order["is_paid"] and order["is_vip"]:
        hold_payment.append(order["item"])
    else:
        canceled.append(order["item"])

print(ship_now, hold_payment, canceled)
```
