# Practice Problem: Online Store Order Dispatcher

## Scenario

You are managing an **online store's dispatch system**. You have a list of orders. Each order is represented as a dictionary containing:

* The item name
* Customer membership status
* Payment confirmation status

You need to loop through the orders and decide whether to:

* **Ship Immediately**
* **Hold Order**
* **Cancel Order**

---

## Starter Code

Copy this starter code directly into your script:

```python
# List of Orders

orders = [
    {"item": "Headphones", "is_paid": True, "is_vip": False},
    {"item": "Keyboard", "is_paid": False, "is_vip": True},
    {"item": "Mousepad", "is_paid": True, "is_vip": True},
    {"item": "Monitor", "is_paid": False, "is_vip": False}
]

# Output Buckets

ship_now = []

hold_payment = []

canceled = []
```

---

## Tasks & Instructions

Write a `for` loop that iterates through each order in `orders`.

Inside the loop, construct an **`if-elif-else`** block.

### Condition 1 — `if`: Ready to Ship

Check if:

```python
order["is_paid"]
```

is **truthy (`True`)**.

If true, append `order["item"]` to the `ship_now` list.

---

### Condition 2 — `elif`: VIP Unpaid Hold

Check if:

```python
order["is_paid"]
```

is **falsy (`False`)** **AND**

```python
order["is_vip"]
```

is **truthy (`True`)**.

If true, append `order["item"]` to the `hold_payment` list.

---

### Condition 3 — `else`: Non-VIP Unpaid Cancellation

Otherwise, the order is an **unpaid non-VIP order**.

Append `order["item"]` to the `canceled` list.

---

## Expected Logic

| Order Status     | Action           |
| ---------------- | ---------------- |
| Paid             | Ship Immediately |
| Unpaid + VIP     | Hold Order       |
| Unpaid + Non-VIP | Cancel Order     |

---

## Output

After the loop finishes, print all three destination lists:

```python
print(ship_now)
print(hold_payment)
print(canceled)
```

### Goal

Write the complete `for` loop and conditional logic yourself. **Don't look for the solution first—try reasoning through the truth values of `is_paid` and `is_vip`.**


# My Python Code:

````python
# List of Orders
orders = [
    {"item": "Headphones", "is_paid": True, "is_vip": False},
    {"item": "Keyboard", "is_paid": False, "is_vip": True},
    {"item": "Mousepad", "is_paid": True, "is_vip": True},
    {"item": "Monitor", "is_paid": False, "is_vip": False}
]

# Output Buckets
ship_now = []
hold_payment = []
canceled = []
for order in orders:
    if order["is_paid"]:
        ship_now.append(order["item"])
    elif not order["is_paid"] and order["is_vip"]:
        hold_payment.append(order["item"])
    else:
        canceled.append(order["item"])

print(ship_now, hold_payment, canceled)
````
