# Practice Problem: Email Notification Filtering System

## Scenario

You are building an **automated notification system** for a platform. You have a list of user message dictionaries.

You need to loop through all messages and categorize them into different email notification lists based on:

* Priority
* Subject availability
* Spam flags

---

## Starter Code

Copy this starter code directly into your script:

```python
# List of Incoming User Messages

messages = [
    {"user": "Alice", "subject": "Urgent: Payment Failed", "is_spam": False},
    {"user": "Bob", "subject": "", "is_spam": False},  # Empty subject string
    {"user": "Charlie", "subject": "Win $1000 NOW!", "is_spam": True},
    {"user": "Diana", "subject": "Weekly Newsletter", "is_spam": False}
]

# Output Buckets

high_priority_emails = []

standard_emails = []

review_folder = []
```

---

## Tasks & Instructions

Write a `for` loop that iterates through each `msg` dictionary inside `messages`.

Inside the loop, construct an **`if-elif-elif-else`** control flow structure.

### Condition 1 — `if`: Spam Check

Check if:

```python
msg["is_spam"]
```

is **truthy (`True`)**.

If true, append `msg["user"]` to the `review_folder` list.

---

### Condition 2 — `elif`: Missing Subject Check

Check if:

```python
msg["subject"]
```

is **falsy** (an empty string `""`).

If true, append `msg["user"]` to the `review_folder` list as well.

---

### Condition 3 — `elif`: Urgent Priority Check

Check if `"Urgent"` is inside:

```python
msg["subject"]
```

If true, append `msg["user"]` to the `high_priority_emails` list.

---

### Condition 4 — `else`: Standard Email

Otherwise, append `msg["user"]` to the `standard_emails` list.

---

## Expected Logic

| Condition                   | Action                |
| --------------------------- | --------------------- |
| Spam                        | Send to Review Folder |
| Missing Subject             | Send to Review Folder |
| Subject contains `"Urgent"` | High Priority Email   |
| Everything else             | Standard Email        |

> **Important:** The order of the conditions matters. Check for spam first, then missing subject, then urgent priority.

---

## Output

Outside the loop, print all three output lists:

```python
print(high_priority_emails)
print(standard_emails)
print(review_folder)
```

## Goal

Write the complete `for` loop and conditional logic yourself.

Think carefully about **truthy/falsy values** and the **order in which the conditions are checked**. 🧠


````python
# List of Incoming User Messages
messages = [
    {"user": "Alice", "subject": "Urgent: Payment Failed", "is_spam": False},
    {"user": "Bob", "subject": "", "is_spam": False},  # Empty subject string
    {"user": "Charlie", "subject": "Win $1000 NOW!", "is_spam": True},
    {"user": "Diana", "subject": "Weekly Newsletter", "is_spam": False}
]

# Output Buckets
high_priority_emails = []
standard_emails = []
review_folder = []

for msg in messages:
    if msg["is_spam"]:
        review_folder.append(msg["user"])
    elif not msg["subject"]:
        review_folder.append(msg["user"])
    elif "Urgent" in msg["subject"]:
        high_priority_emails.append(msg["user"])
    else:
        standard_emails.append(msg["user"])

print(high_priority_emails)
print(standard_emails)
print(review_folder)
````
