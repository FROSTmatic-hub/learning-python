# My Python Code:

````python
cart = ["Shirt", "Pants", "Trousers", "Shirt", "Shoes"]
cart.pop(4)  #removes shoes which is at index 4
cart.append("Glasses") # Adds Glasses to the end of the cart list

tuple = ("Alex", "New York")
name, city = tuple

product_prices = {
    cart[0] : 180,
    cart[1] : 250,
    cart[2] : 300,
    cart[3] : 180,
    cart[4] : 100
}

unique_cart = set(cart)
discount_product = {"Shirt", "Glasses"}
common_products = unique_cart.intersection(discount_product)

total_price = product_prices[cart[0]] + product_prices[cart[1]] + product_prices[cart[2]] + product_prices[cart[3]] + product_prices[cart[4]]

summary = f"Customer: {name}\nCity: {city}\n\nCart: {cart}\nUnique Products: {unique_cart}\n\nDiscounted Products: {discount_product}\nProducts on Discount: {common_products}\n\nTotal: {total_price}\n"

print(summary)






````
