# Write a Python program that creates a small user profile and analyzes some of its data.

## Your program should:

* Create a dictionary containing:
  * `username`
  * `age`
  * `email`
  * `skills` → a list containing at least 4 skills, with at least one duplicate
  * `location`
* Extract the `skills` list from the dictionary.
* Create a set from the `skills` list to remove duplicates.
* Create a variable containing the number of unique skills.
* Create a variable called `account_status` containing either:
  * `"Active"`
  * `"Inactive"`
* Use `.get()` to safely retrieve a dictionary value that does not exist, and store the result in a variable.
* Create a formatted summary using an f-string.

### Your output should communicate something similar to:

```text
Username: ...
Age: ...
Email: ...
Location: ...
Skills: [...]
Unique Skills: [...]
Number of Unique Skills: ...
Account Status: ...
Missing Information: None

# My Python Code:

````python
user_info = {
    "Username" : "Alex",
    "Age" : 24,
    "Email" : "alex123@gmail.com",
    "Skills" : ["Programming", "Writing", "Editing", "Writing"],
    "Location" : "New York"
}

extracted_skills = user_info.get('Skills')
unique_skills = set(extracted_skills)
num_skills = len(unique_skills)

account_status = "Active"

missing_info = user_info.get('Place')

summary = f"Username: {user_info['Username']}\nAge: {user_info['Age']}\nEmail: {user_info['Email']}\nLocation: {user_info['Location']}\nSkills: {user_info['Skills']}\nUnique Skills: {unique_skills}\nNumber of Unique Skills: {num_skills}\nAccount Status: {account_status}\nMissing Information: {missing_info}\n"
print(summary)
````
