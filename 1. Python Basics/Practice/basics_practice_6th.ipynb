# Level 1 — Challenge 5: Mini Inventory System

## Scenario
- You're building a tiny inventory system for a computer shop.
- Create a program with the following data:
`products = ["Laptop", "Mouse", "Keyboard", "Monitor", "Mouse"]`

- Your program must:
1. Add "Headphones" to the inventory.
2. Remove one occurrence of "Mouse".
3. Create a set from the inventory to get the unique products.
4. Create a dictionary containing the price of every unique product. You choose the prices.
5. Create a tuple containing:
`("Tech Store", "Aligarh")`
Unpack it into:
`store_name`
`city`
6. Create a set of products currently on sale. Choose at least 2 products.
7. Use a set operation to determine which unique inventory products are currently on sale.
8. Calculate the total value of the inventory.
    - Important: Because `"Mouse"` originally appeared twice and you only removed one occurrence, if one `"Mouse"` remains, its price must be counted.
    - You cannot use loops yet, so calculate the total explicitly.

9. Create a dictionary called store_info containing:
```python
store_name
city
total_products
unique_products
inventory_value
```
10. Use `.get()` to safely retrieve a key called `"owner"` that doesn't exist.
11. Print a formatted store report.
```python
===== TECH STORE =====

Store: ...
City: ...

Inventory: [...]
Unique Products: [...]

Products on Sale: [...]
Sale Products in Inventory: [...]

Total Products: ...
Inventory Value: ...
Owner: None
```

# My Python Code:

````python
products = ["Laptop", "Mouse", "Keyboard", "Monitor", "Mouse"]

# 1. Adding "headphones"
products.append("Headphones")

# 2. removing one occurence of "Mouse"
products.remove("Mouse")

# 3. creating a set 
unique_products = set(products)
# {'Monitor', 'Mouse', 'Laptop', 'Keyboard', 'Headphones'}

# 4. creating a dictionary
product_price = {
    "Monitor" : 3000,
    "Mouse" : 400,
    "Laptop" : 50000,
    "Keyboard" : 1200,
    "Headphones" : 700
}

# 5. creating a tuple
store_data = ("Tech Store", "Aligarh")
store_name, city = store_data

# 6. creating a set of product on sale
sale_product = {"Laptop", "Monitor"}

# 7. using set operation
sale_products_in_inventory = unique_products.intersection(sale_product)

# 8. calculating total price
total_value = product_price["Monitor"] + product_price["Headphones"] + product_price["Keyboard"] + product_price["Laptop"] + product_price["Mouse"]

# 9. creating another dict
store_info = {
    "store_name" : store_name,
    "city" : city,
    "total_products" : len(products),
    "unique_products" : unique_products,
    "value" : total_value
}
# 10. using get to retrive "owner"
owner = store_info.get('owner')

# 11. formatted report
summary = f"===== TECH STORE =====\n\nStore: {store_name}\nCity: {city}\n\nInventory: {products}\nUnique Products: {unique_products}\n\nProducts on Sale: {sale_product}\n Sale Products in Inventory: {sale_products_in_inventory}\n\nTotal Products: {len(products)}\nInventory Value: {total_value}\nOwner: {owner}"

print(summary)
````
