# Level 1 — Challenge 1: Student Profile

Write a **Python** program that:

*   Creates variables for:
    *   Student's name
    *   Age
    *   Three subject marks
*   Calculates the total marks and average marks.
*   Stores the subjects and marks in a dictionary.
*   Creates a list containing the subject names.
*   Creates a set from the subject names to demonstrate uniqueness.
*   Prints a formatted summary like:
    > Student: Shaafi
    >
    > Age: 20
    >
    > Marks: {'Python': 85, 'Linux': 78, 'Networking': 92}
    >
    > Total: 255
    >
    > Average: 85.0
    >
    > Subjects: ['Python', 'Linux', 'Networking']
    >
    > Unique Subjects: {'Python', 'Linux', 'Networking'}

# My Python Code:

```python
student_name = "Alex"
student_age = 24
student_marks = {
    "English" : 78,
    "Maths" : 89,
    "Science" : 81
}

total_marks = student_marks["English"] + student_marks["Maths"] + student_marks["Science"]
average_marks = total_marks / 3

subject_names = ["English", "Maths", "Science"]
unique_subject = set(subject_names)

summary = f"Student: {student_name}\nAge: {student_age}\nMarks: {student_marks}\nTotal: {total_marks}\nAverage: {average_marks}\nSubjects: {subject_names}\nUnique Subjects: {unique_subject}\n"
print(summary)
```
