# Problem Title: Cross-Platform Threat Intel & Privilege Access Audit

You are tasked with auditing an enterprise security log containing permission badges across multiple environments. You need to verify privilege escalations, eliminate duplicates, inspect sub-access levels, and produce a final set-state verification report.

## Input Data

Copy this starter code directly into your script:

```python
# Raw access logs with duplicates and inconsistent user badges

network_admin_badges = ["read", "write", "execute", "firewall", "write", "audit"]

cloud_infra_badges = ["write", "deploy", "firewall", "backup", "deploy"]

incident_response_badges = ["monitor", "isolate", "dump_memory", "read"]

mandatory_baseline = {"read", "audit"}
```

## Step-by-Step Requirements

Every single operation below tests a specific set property or method covered in Sections 1.25 & 1.25.1.

### 1. Initialization, Deduplication, & Membership Testing

* Create `network_set`, `cloud_set`, and `ir_set` from their respective raw lists.
* Add `"sudo"` to `network_set` using `.add()`.
* Add `"audit"` to `network_set` again using `.add()` to verify duplicate rejection.
* In a single line, check if `"audit"` exists in `network_set` using the `in` operator, and print the resulting boolean.
* In a single line using a comment, explain what happens if you attempt to access `network_set[0]`.

### 2. Independence via Copying

* Create an isolated backup of `network_set` named `audit_vault` using `.copy()`.
* Remove all permissions inside `audit_vault` using `.clear()`.
* Verify `network_set` was not affected by printing `len(network_set)` and `len(audit_vault)`.

### 3. Pure Comparisons (Non-Modifying Operations)

**Intersection (`&` / `.intersection()`):**

Find badges present in both `network_set` and `cloud_set`. Store the result in `shared_infra`.

**Union (`|` / `.union()`):**

Combine all unique permissions from `cloud_set` and `ir_set` without duplicates into `field_ops`.

**Difference (`.difference()`):**

Identify all badges exclusive to `network_set` that are absent from `cloud_set`. Store this new set in `exclusive_net`.

* Confirm that `network_set` itself remains completely unchanged after calling `.difference()`.

### 4. In-Place Modifications & Removal Methods

* Safely remove `"firewall"` from `cloud_set` using `.discard()`.
* Call `.discard("quantum_key")` on `cloud_set` to prove that discarding an absent element does not trigger an error.
* Perform an in-place difference update on `network_set`: remove every element found in `ir_set` directly using `.difference_update()`.

### 5. Boolean Relationship Analysis

**Disjoint Check (`.isdisjoint()`):**

Check if `cloud_set` and `ir_set` are completely disjoint. Store the boolean in `is_ops_separated`.

**Subset Check (`.issubset()`):**

Check whether `mandatory_baseline` is a valid subset of `exclusive_net`. Store the boolean in `has_baseline`.

**Superset Check (`.issuperset()`):**

Check whether `network_set` is a superset of `{"write", "sudo"}`. Store the boolean in `is_super_admin`.

### 6. Final Report Generation

Print all generated sets and boolean flags.


````python
network_admin_badges = ["read", "write", "execute", "firewall", "write", "audit"]
cloud_infra_badges = ["write", "deploy", "firewall", "backup", "deploy"]
incident_response_badges = ["monitor", "isolate", "dump_memory", "read"]
mandatory_baseline = {"read", "audit"}

# 1. Initilization
network_set = set(network_admin_badges)
cloud_set = set(cloud_infra_badges)
ir_set = set(incident_response_badges)

#2. copying and clearing
network_set.add("sudo") 
network_set.add("audit")
print("audit" in network_set)
# print(network_set[0]) If we try to access network_set[0], we get an type error because sets cannot be accessed using indexes.
audit_vault = network_set.copy()
audit_vault.clear()
print(len(network_set))
print(len(audit_vault))

#3. Interection
shared_infra = network_set & cloud_set
print(shared_infra)

#Union
field_ops = cloud_set.union(ir_set)
print(field_ops)

#Difference
exclusive_net = network_set.difference(cloud_set)
print(exclusive_net)

#confirming network_set
print(network_set)

#4. Modifications
cloud_set.discard("firewall")
print(cloud_set)

print(cloud_set.discard("quantum_key")) #This prints None

network_set.difference_update(ir_set)
print(network_set)
print(ir_set)

#5. Boolean 
is_ops_separated = cloud_set.isdisjoint(ir_set)
print(is_ops_separated)

has_baseline = mandatory_baseline.issubset(exclusive_net)
print(has_baseline)

is_super_admin = network_set.issuperset({"write", "sudo"})
print(is_super_admin)
````
